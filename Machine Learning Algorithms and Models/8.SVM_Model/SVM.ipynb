```python
#Load the Dataset
df=pd.read_csv("/content/bank.csv", delimiter=';')
print("Dataset Loaded")
```

```python
#Find Categorical and Numerical columns
numerical_cols=df.select_dtypes(include=['int64']).columns
categorical_cols=df.select_dtypes(include=['object']).columns
print("Numerical Columns: ",numerical_cols)
print("Categorical Columns: ",categorical_cols)
```

```python
#Separete Features and target values
x=df.iloc[:,:-1]
y=df.iloc[:,-1]
y = y.map({'no': 0, 'yes': 1})
```

```python
#split the data into train and test
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
#Identify categorical data of x
categorical_x=x.select_dtypes(include=['object','category']).columns
print(categorical_x)
```

```python
#Apply one-hot encoding
x_test=pd.get_dummies(x_test, columns=categorical_x, drop_first=True)
x_train=pd.get_dummies(x_train, columns=categorical_x, drop_first=True)

x_train,x_test=x_train.align(x_test,join='outer',axis=1, fill_value=0)
```

```python
#standardization
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

```python
#Model Building
from sklearn.svm import LinearSVC
model=LinearSVC(random_state=42)
model.fit(x_train,y_train)
```

```python
#Predictions
y_pred=model.predict(x_test)
accuracy=accuracy_score(y_test,y_pred)
print("Accuracy: ",accuracy)
```

```python
#Evalution Metrices
confusion_mat=confusion_matrix(y_test,y_pred)
print("Confusion Matrix : ",confusion_mat)
```

```python
#classification report
classification_rep=classification_report(y_test,y_pred)
print("Classification Report:")
print(classification_rep)
```
